```python

def alpha_beta(node, depth, alpha, beta, maximizing):
    if isinstance(node, int):
        return node

    if maximizing:
        value = float('-inf')

        for i, child in enumerate(node):
            value = max(value, alpha_beta(
                child, depth + 1, alpha, beta, False
            ))

            alpha = max(alpha, value)

            if alpha >= beta:
                print("Pruned subtree:", node[i+1:])
                break

        return value

    else:
        value = float('inf')

        for i, child in enumerate(node):
            value = min(value, alpha_beta(
                child, depth + 1, alpha, beta, True
            ))

            beta = min(beta, value)

            if beta <= alpha:
                print("Pruned subtree:", node[i+1:])
                break

        return value


# Game tree from the PPT
tree = [
    [[10, 9], [14, 18]],
    [[5, 4], [50, 3]]
]

result = alpha_beta(
    tree, 0, float('-inf'), float('inf'), True
)

print("Final MAX value:", result)
```
