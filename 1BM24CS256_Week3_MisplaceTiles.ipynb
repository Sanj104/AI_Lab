```python
import heapq


initial = (1, 2, 3,
           4, 0, 6,
           7, 5, 8)


goal = (1, 2, 3,
        4, 5, 6,
        7, 8, 0)



def misplaced_tiles(state):
    count = 0

    for i in range(9):
        if state[i] != 0 and state[i] != goal[i]:
            count += 1

    return count



def print_puzzle(state):
    for i in range(0, 9, 3):
        print(state[i], state[i + 1], state[i + 2])
    print()



def get_neighbors(state):
    neighbors = []

    zero = state.index(0)
    row = zero // 3
    col = zero % 3

    moves = [
        (-1, 0, "UP"),
        (1, 0, "DOWN"),
        (0, -1, "LEFT"),
        (0, 1, "RIGHT")
    ]

    for dr, dc, move in moves:
        new_row = row + dr
        new_col = col + dc

        if 0 <= new_row < 3 and 0 <= new_col < 3:

            new_zero = new_row * 3 + new_col

            new_state = list(state)
            new_state[zero], new_state[new_zero] = \
                new_state[new_zero], new_state[zero]

            neighbors.append((tuple(new_state), move))

    return neighbors



def a_star():


    pq = []

    h = misplaced_tiles(initial)

    heapq.heappush(pq, (h, 0, initial, []))

    visited = set()

    while pq:

        f, g, state, path = heapq.heappop(pq)

        if state in visited:
            continue

        visited.add(state)

        print("g =", g,
              "h =", misplaced_tiles(state),
              "f =", f)

        print_puzzle(state)


        if state == goal:
            print("Goal Reached!")
            print("Moves:", path)
            print("Number of moves:", len(path))
            return


        for new_state, move in get_neighbors(state):

            if new_state not in visited:

                new_g = g + 1
                new_h = misplaced_tiles(new_state)
                new_f = new_g + new_h

                heapq.heappush(
                    pq,
                    (new_f, new_g, new_state, path + [move])
                )


a_star()
```
