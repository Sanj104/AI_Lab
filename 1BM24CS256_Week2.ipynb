```python
# 8 Puzzle using DFS
# Maximum depth = 10

initial = (5, 8, 2,
           1, 0, 3,
           4, 7, 6)

goal = (1, 2, 3,
        4, 5, 6,
        7, 8, 0)

MAX_DEPTH = 10


def print_puzzle(state):
    for i in range(0, 9, 3):
        print(state[i], state[i + 1], state[i + 2])
    print()


def get_neighbors(state):

    neighbors = []

    blank = state.index(0)

    row = blank // 3
    col = blank % 3

    moves = [
        (-1, 0, "Up"),
        (1, 0, "Down"),
        (0, -1, "Left"),
        (0, 1, "Right")
    ]

    for dr, dc, move in moves:

        new_row = row + dr
        new_col = col + dc

        if 0 <= new_row < 3 and 0 <= new_col < 3:

            new_blank = new_row * 3 + new_col

            new_state = list(state)

            new_state[blank], new_state[new_blank] = \
                new_state[new_blank], new_state[blank]

            neighbors.append((tuple(new_state), move))

    return neighbors


def dfs(state, path, moves, depth):

    # Goal reached
    if state == goal:
        return path, moves

    # Stop after 10 moves
    if depth == MAX_DEPTH:
        return None

    for new_state, move in get_neighbors(state):

        # Avoid repeating states in current path
        if new_state not in path:

            path.append(new_state)
            moves.append(move)

            result = dfs(
                new_state,
                path,
                moves,
                depth + 1
            )

            if result is not None:
                return result

            # Backtracking
            path.pop()
            moves.pop()

    return None


# Start DFS
result = dfs(initial, [initial], [], 0)


if result:

    path, moves = result

    print("DFS Solution Found")
    print("Number of moves:", len(moves))
    print("Moves:", moves)
    print()

    for i, state in enumerate(path):

        print("Step", i)
        print_puzzle(state)

else:

    print("Goal cannot be reached within 10 moves.")
```

```python
# 8 Puzzle using Iterative Deepening Search (IDS)

initial = (1, 5, 2,
           4, 8, 3,
           7, 0, 6)

goal = (1, 2, 3,
        4, 5, 6,
        7, 8, 0)


def print_puzzle(state):
    for i in range(0, 9, 3):
        print(state[i], state[i + 1], state[i + 2])
    print()


def get_neighbors(state):

    neighbors = []

    blank = state.index(0)

    row = blank // 3
    col = blank % 3

    moves = [
        (-1, 0, "Up"),
        (1, 0, "Down"),
        (0, -1, "Left"),
        (0, 1, "Right")
    ]

    for dr, dc, move in moves:

        new_row = row + dr
        new_col = col + dc

        if 0 <= new_row < 3 and 0 <= new_col < 3:

            new_blank = new_row * 3 + new_col

            new_state = list(state)

            new_state[blank], new_state[new_blank] = \
                new_state[new_blank], new_state[blank]

            neighbors.append((tuple(new_state), move))

    return neighbors


def depth_limited_search(state, path, moves, depth_limit):

    # Goal reached
    if state == goal:
        return path, moves

    # Depth limit reached
    if depth_limit == 0:
        return None

    # Generate possible moves
    for new_state, move in get_neighbors(state):

        # Avoid repeating states in current path
        if new_state not in path:

            path.append(new_state)
            moves.append(move)

            result = depth_limited_search(
                new_state,
                path,
                moves,
                depth_limit - 1
            )

            if result is not None:
                return result

            # Backtracking
            path.pop()
            moves.pop()

    return None


def ids(initial):

    depth = 0

    while True:

        print("Searching at depth:", depth)

        result = depth_limited_search(
            initial,
            [initial],
            [],
            depth
        )

        if result is not None:
            return result

        depth += 1


# Run IDS
path, moves = ids(initial)


print("\nIDS Solution Found")
print("Number of moves:", len(moves))
print("Moves:", moves)

print("\nSolution Path:")

for i, state in enumerate(path):

    print("Step", i)
    print_puzzle(state)
```
