```python

import random
import math

N = 8

# Calculate number of attacking queen pairs
def calculate_cost(board):
    cost = 0

    for i in range(N):
        for j in range(i + 1, N):
            if (board[i] == board[j] or
                abs(board[i] - board[j]) == abs(i - j)):
                cost += 1

    return cost


# Simulated Annealing algorithm
def simulated_annealing():
    random.seed(2)

    # Generate initial random arrangement
    current = [random.randint(0, 7) for _ in range(N)]

    temperature = 100
    cooling_rate = 0.99
    iterations = 0

    while temperature > 0.1:
        current_cost = calculate_cost(current)

        if current_cost == 0:
            break

        # Generate a random neighbour
        next_state = current.copy()

        col = random.randint(0, 7)
        row = random.randint(0, 7)

        while row == next_state[col]:
            row = random.randint(0, 7)

        next_state[col] = row

        next_cost = calculate_cost(next_state)

        delta = current_cost - next_cost

        # Accept or reject the new state
        if delta > 0:
            current = next_state
        else:
            probability = math.exp(delta / temperature)

            if random.random() < probability:
                current = next_state

        temperature *= cooling_rate
        iterations += 1

    return current, calculate_cost(current), iterations


# Display solution
solution, cost, steps = simulated_annealing()

print("Simulated Annealing - 8 Queens")
print("Final Board Arrangement:", solution)
print("Minimum Cost:", cost)
print("Iterations:", steps)

if cost == 0:
    print("Solution Found!")
else:
    print("Solution Not Found.")

print("\nChessboard:")

for row in range(N):
    for col in range(N):
        if solution[col] == row:
            print(" Q ", end="")
        else:
            print(" . ", end="")
    print()
```
